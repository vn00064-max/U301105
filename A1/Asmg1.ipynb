```python
# Identify the top 10 cities with the highest number of restaurants: New Delhi, Gurgaon, Nodia, Faridabad, Ghaziabad, Ahmedabad, Amristar, Bhubaneswar
# Guwanhati, Lucknow
top_cities = df["City"].value_counts().head(10)
print(top_cities)
```

```python
# ----------------------------------------------------------------------------------------------------------------
# Part 3 - Visualizations
# ----------------------------------------------------------------------------------------------------------------
# Create appropriate visualizations using Matplotlib and/or Seaborn.
```

```python
# Bar Chart: Top 10 cities with the highest number of restaurants.
plt.figure(figsize=(10,5))

sns.barplot(
    x=top_cities.index,
    y=top_cities.values
)

plt.title("Top 10 Cities by Restaurant Count")
plt.xlabel("City")
plt.ylabel("Number of Restaurants")

plt.xticks(rotation=45)

plt.show()
```

```python
# Histogram: Distribution of Aggregate Ratings
plt.figure(figsize=(8,5))

plt.hist(
    df["Aggregate rating"],
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Aggregate Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")

plt.show()
```

```python
# Bar Chart: Number of restaurants offering Online Delivery (Yes/No).
delivery_counts = df["Has Online delivery"].value_counts()

sns.barplot(
    x=delivery_counts.index,
    y=delivery_counts.values
)

plt.title("Online Delivery Availability")
plt.xlabel("Online Delivery")
plt.ylabel("Restaurant Count")

plt.show()
```

```python
top_cuisines = df["Cuisines"].value_counts().head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=top_cuisines.values, 
    y=top_cuisines.index
)

plt.title("Top 10 Most Common Cuisines")
plt.xlabel("Number of Restaurants")
plt.ylabel("Cuisine")

plt.tight_layout()
plt.show()
```

```python
#  Line chart: Average restaurant rating by Price Range
price_rating = df.groupby("Price range")["Aggregate rating"].mean().reset_index()
plt.figure(figsize=(8,5))
sns.lineplot(
    data=price_rating,
    x="Price range",
    y="Aggregate rating",
    marker="o",
    linewidth=2
)

plt.title("Average Rating by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Average Rating")

plt.show()
```
